# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]
labels = [f'Doc {i+1}' for i in range(len(sentences))]

# 1) Convert the sentences to TF-IDF vectors
tfidf_vec = TfidfVectorizer()
vectors = tfidf_vec.fit_transform(sentences)

# 2) Cosine similarity between every pair of sentences
sim_matrix = cosine_similarity(vectors)
df_sim = pd.DataFrame(sim_matrix, index=labels, columns=labels).round(3)
df_sim
````

````python
# Find the most similar pair of different sentences
best = None
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        if best is None or sim_matrix[i][j] > best[0]:
            best = (sim_matrix[i][j], i, j)

print(f"Most similar pair: {labels[best[1]]} and {labels[best[2]]} (similarity = {best[0]:.3f})")
print(sentences[best[1]])
print(sentences[best[2]])
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
docs = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]
doc_labels = [f'Doc {i+1}' for i in range(len(docs))]

tfidf_2 = TfidfVectorizer()
tfidf_matrix = tfidf_2.fit_transform(docs)

df_tfidf = pd.DataFrame(tfidf_matrix.toarray(),
                        columns=tfidf_2.get_feature_names_out(),
                        index=doc_labels).round(3)
df_tfidf
````

````python
# The most important words in each document (highest TF-IDF scores)
for label, row in df_tfidf.iterrows():
    top = row.sort_values(ascending=False).head(3)
    print(label, '->', ', '.join(f"{w} ({s})" for w, s in top.items()))

# The most important words over all the documents (sum of TF-IDF scores)
print()
print(df_tfidf.sum().sort_values(ascending=False).head(5))
````

````python
import os
import re
import glob
import pandas as pd
import gensim
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

# Use the local file if it is next to the notebook, otherwise download a copy from Kaggle
CSV_NAME = 'simpsons_script_lines.csv'
if os.path.exists(CSV_NAME):
    csv_path = CSV_NAME
else:
    import kagglehub
    folder = kagglehub.dataset_download("pierremegret/dialogue-lines-of-the-simpsons")
    csv_path = glob.glob(os.path.join(folder, '**', '*.csv'), recursive=True)[0]
print('Using file:', csv_path)

simpsons = pd.read_csv(csv_path, low_memory=False)
print(simpsons.shape)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

# Use the 'spoken_words' column (drop the empty lines first)
simpsons = simpsons.dropna(subset=['spoken_words'])
simpsons['clean_text'] = simpsons['spoken_words'].astype(str).apply(clean_text)

# Tokenize every line
tokens = [word_tokenize(line) for line in simpsons['clean_text']]
tokens = [t for t in tokens if len(t) > 0]
print('Number of lines:', len(tokens))
print(tokens[0])
````

````python
# Skip-gram Word2Vec model
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1)
print('Vocabulary size:', len(Skip_gram_model.wv))
````

### Task 4

````python
print("Words similar to 'homer':")
for word, score in Skip_gram_model.wv.most_similar('homer'):
    print(f"  {word:15s} {score:.4f}")

print("\nWords similar to 'marge':")
for word, score in Skip_gram_model.wv.most_similar('marge'):
    print(f"  {word:15s} {score:.4f}")

print("\nWords similar to 'bart':")
for word, score in Skip_gram_model.wv.most_similar('bart'):
    print(f"  {word:15s} {score:.4f}")
````
